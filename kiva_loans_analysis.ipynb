# Question 1: Which sector received the highest total funded amount?

Agriculture received the highest total funded amount, about 134 million dollars, ahead of Food with about 115 million. Here funded_amount is the money actually raised, not the amount requested. This may be because Agriculture has the most loans (180,302), rather than larger individual loans.

# Question 2: Do loans whose purpose mentions 'flour' last longer or shorter than other loans?

This might be because businesses that buy baking ingredients can sell their products and recover the money quickly the loan term is shorter

# Question 3: If I label each loan as short or long based on 8 months, how do the labels split across countries?

I labelled a loan as "short" if it lasts less than 8 months and "long" otherwise, because only around 11 per cent of loans are ‘short’ (75,295 out of 671,205). Overall, 75,295 loans are short and 595,910 are long.

Among the countries with the most loans, Perù has the highest share of short loans (35%), while Pakistan has the lowest (0%).

However, countries with few loans have percentages that are not very reliable.

# Question 4: How many total lenders does Nicaragua attract?

Loans in Nicaragua attract a total of 280,818 lenders, an average of 23.8 per loan, which is slightly higher than the 20.6 average across the whole dataset.

# Question 5: How many different countries have loans above 750 dollars?

87 out of 87 countries have at least one loan above 750, making a total of 216,520 loans.
I used 750 as it lies between the median (500) and the 75th percentile (1,000), and I used ‘>’, so loans of exactly 750 are excluded
Every country has large loans, so the threshold does not distinguish between countries

# Question 6: Which are the 10 countries with the highest levels of borrowing, and how much do they borrow on average?

The Philippines tops the list with 160,441 loans, well ahead of Kenya with 75,825, but it has the lowest average loan amount (about 345 dollars), whereas Peru has the highest (about 1,388 dollars).

```python
df["term_label"] = df["term_in_months"].apply(lambda x: "short" if x < 8 else "long")
```

```python
df["term_label"].value_counts()
```

```python
pd.crosstab(df["country"], df["term_label"], normalize="index")
```

```python
shares = pd.crosstab(df["country"], df["term_label"], normalize="index")
```

```python
shares.loc[["Philippines", "Kenya", "El Salvador", "Cambodia", "Pakistan", "Peru", "Colombia", "Uganda", "Tajikistan", "Ecuador"]].sort_values("short", ascending=False)
```
